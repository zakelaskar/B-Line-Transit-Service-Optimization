## Data Cleaning for CSV and Excel Files
This block loads all `.csv` and `.xlsx` files from `data/raw/`, cleans them using `clean_dataframe()`, and saves the cleaned version to `data/processed/`.

```python
import os
import pandas as pd

# Paths
raw_dir = "C:/Users/zelaskar/Documents/B-Line_Project/data/raw"
processed_dir = "C:/Users/zelaskar/Documents/B-Line_Project/data/processed"

# Helper: Clean a dataframe
def clean_dataframe(df):
    df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')

    if 'timestamp' in df.columns:
        df['timestamp'] = pd.to_datetime(df['timestamp'], errors='coerce')

    # Drop empty columns
    df = df.dropna(axis=1, how='all')
    
    return df

# Process each file
for filename in os.listdir(raw_dir):
    file_path = os.path.join(raw_dir, filename)

    try:
        # Load file
        if filename.endswith(".csv"):
            df = pd.read_csv(file_path)
        elif filename.endswith(".xlsx"):
            df = pd.read_excel(file_path, sheet_name=0)
        else:
            print(f"Skipping unsupported file: {filename}")
            continue

        # Clean it
        df_cleaned = clean_dataframe(df)

        # Save cleaned version
        new_filename = filename.replace(".csv", "_cleaned.csv").replace(".xlsx", "_cleaned.csv")
        output_path = os.path.join(processed_dir, new_filename)
        df_cleaned.to_csv(output_path, index=False)

        print(f"✔ Cleaned and saved: {new_filename}")

    except Exception as e:
        print(f"❌ Error processing {filename}: {e}")
```

## TXT File Cleaning
This block is for processing `.txt` fare logs (if any). Can be skipped if no TXT files exist.

```python
import os
import pandas as pd

# Set your paths
txt_raw_dir = "C:/Users/zelaskar/Documents/B-Line_Project/data/raw"
processed_dir = "C:/Users/zelaskar/Documents/B-Line_Project/data/processed"

# Loop through and clean only .txt files
for filename in os.listdir(txt_raw_dir):
    if filename.endswith(".txt"):
        file_path = os.path.join(txt_raw_dir, filename)
        print(f"Processing: {filename}")

        try:
            # Load tab-separated .txt file
            df = pd.read_csv(file_path, sep="\t", encoding="utf-8")

            # Clean column names
            df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

            # Parse datetime column (if available)
            if 'date_time' in df.columns:
                df['date_time'] = pd.to_datetime(df['date_time'], errors='coerce')

            # Drop empty columns
            df = df.dropna(axis=1, how="all")

            # Save cleaned file
            new_filename = filename.replace(".txt", "_cleaned.csv")
            output_path = os.path.join(processed_dir, new_filename)
            df.to_csv(output_path, index=False)

            print(f"✔ Saved cleaned file: {new_filename}")

        except Exception as e:
            print(f"❌ Error processing {filename}: {e}")
```

## Data Catalog

```python
import os
import pandas as pd

# Set your processed folder path
processed_dir = "C:/Users/zelaskar/Documents/B-Line_Project/data/processed"
catalog = []

# Loop through all cleaned .csv files
for filename in os.listdir(processed_dir):
    if filename.endswith(".csv"):
        file_path = os.path.join(processed_dir, filename)
        try:
            # Load full file (or sample if very large)
            df = pd.read_csv(file_path)

            # Identify columns with more than 10% missing values
            missing_pct = df.isnull().mean()
            high_missing_cols = missing_pct[missing_pct > 0.10].index.tolist()

            info = {
                "Filename": filename,
                "Type": (
                    "Validation" if "validation" in filename.lower() else
                    "Transaction" if "transaction" in filename.lower() else
                    "Ridership" if "ridership" in filename.lower() else
                    "Other"
                ),
                "Rows": df.shape[0],
                "Columns": df.shape[1],
                "Column Names": ", ".join(df.columns[:5]) + ("..." if len(df.columns) > 5 else ""),
                "Has Timestamp": 'timestamp' in df.columns,
                "Time Range": (
                    f"{df['timestamp'].min()} to {df['timestamp'].max()}"
                    if 'timestamp' in df.columns else "N/A"
                ),
                "Columns With >10% Missing": ", ".join(high_missing_cols) if high_missing_cols else "None"
            }
        except Exception as e:
            info = {
                "Filename": filename,
                "Type": "Unreadable",
                "Rows": "N/A",
                "Columns": "N/A",
                "Column Names": f"Error: {str(e)}",
                "Has Timestamp": False,
                "Time Range": "N/A",
                "Columns With >10% Missing": "N/A"
            }

        catalog.append(info)

# Create and save catalog
catalog_df = pd.DataFrame(catalog)
catalog_path = os.path.join(processed_dir, "data_catalog.csv")
catalog_df.to_csv(catalog_path, index=False)

# View result
catalog_df
```
